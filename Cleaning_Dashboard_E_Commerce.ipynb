```python
import pandas as pd
```

```python
d1 = pd.read_csv('ecommerce - event_data.csv')
```

```python
d1.head(10)
```

```python
len(d1)
```

```python
d1_type = d1.dtypes
print(d1_type)
```

```python
# Identifying null values variable
isnull = d1.isnull()

# Counting null values by Column
isnull_count = isnull.sum()

print("Number of Nulls:", isnull_count)

# Total Nulls in Dataset
Total_isnull_count = isnull.sum().sum()

print("Total Nulls:", isnull.sum().sum())
```

```python
# Cari baris yang duplikat
duplikat = d1[d1.duplicated()]

print("Baris yang duplikat:")
print(d1[duplikat])
```

```python
d2 = d1.drop_duplicates()
```

```python
len(d2)
```

```python
# Lihat jumlah data kosong per kolom
print("Jumlah data kosong per kolom:")
print(d1.isnull().sum())
```

```python
d3 = d1.dropna()
```

```python
len(d3)
```

```python
d3.head(10)
```

```python
# Identifying null values variable
isnull = d3.isnull()

# Counting null values by Column
isnull_count = isnull.sum()

print("Number of Nulls:", isnull_count)

# Total Nulls in Dataset
Total_isnull_count = isnull.sum().sum()

print("Total Nulls:", isnull.sum().sum())
```

```python
d3.to_csv('ecommerce_cleaned.csv', index=False)
```
